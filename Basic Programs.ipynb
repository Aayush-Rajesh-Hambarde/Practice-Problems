```python
num1 = float(input("Enter first number: "))
print("The first number is:",num1)
num2 = float(input("Enter second number: "))
print("The second number is:",num2)

print("1. Addition")
print("2. Subtraction")
print("3. Multiplication")
print("4. Division")

choice = int(input("Enter your choice (1-4): "))
print("The choice entered is:",choice)

if choice == 1:
    print("Result =", num1 + num2)
elif choice == 2:
    print("Result =", num1 - num2)
elif choice == 3:
    print("Result =", num1 * num2)
elif choice == 4:
    if num2 != 0:
        print("Result =", num1 / num2)
    else:
        print("Error: Division by zero")
else:
    print("Invalid choice")
```

```python
num = int(input("Enter a number: "))
print("The number is:",num)
sum = 0

while num > 0:
    digit = num % 10
    sum = sum + digit
    num = num // 10

print("Sum of digits =", sum)
```

```python
num = int(input("Enter a number: "))
print("The number is:",num)
fact = 1

if num < 0:
    print("Factorial does not exist for negative numbers")
else:
    for i in range(1, num + 1):
        fact = fact * i
    print("Factorial of", num, "is", fact)
```

```python
num = int(input("Enter a number: "))
print("The number is:",num)

if num <= 1:
    print(num, "is not a Prime number")
else:
    is_prime = True
    for i in range(2, int(num ** 0.5) + 1):
        if num % i == 0:
            is_prime = False
            break
    if is_prime:
        print(num, "is a Prime number")
    else:
        print(num, "is not a Prime number")
```

```python
print("Prime numbers between 10 and 99 are:")

for num in range(10, 100):
    is_prime = True
    for i in range(2, int(num ** 0.5) + 1):
        if num % i == 0:
            is_prime = False
            break
    if is_prime:
        print(num, end=" ")
```

```python
num = int(input("Enter a number: "))
print("The number entered is:",num)
reverse = 0

while num > 0:
    digit = num % 10
    reverse = reverse * 10 + digit
    num = num // 10

print("Reversed number =", reverse)
```

```python
year = int(input("Enter a year: "))
print("The year entered is:",year)

if (year % 4 == 0 and year % 100 != 0) or (year % 400 == 0):
    print(year, "is a Leap Year")
else:
    print(year, "is not a Leap Year")
```

```python
marks = float(input("Enter marks in Math: "))
print("Marks entered:",marks)

if marks >= 90 and marks <= 100:
    grade = "O"
elif marks >= 80 and marks <= 89:
    grade = "A+"
elif marks >= 70 and marks <= 79:
    grade = "A"
elif marks >= 60 and marks <= 69:
    grade = "B"
elif marks >= 50 and marks <= 59:
    grade = "C"
elif marks >= 40 and marks <= 49:
    grade = "P"
elif marks < 40:
    grade = "F"
else:
    grade = "Invalid marks"

print("Grade assigned:", grade)
```

```python
name = input("Enter Student Name: ")
print("Student name entered:",name)
roll = input("Enter Roll Number: ")
print("Roll number entered:",roll)

math = float(input("Enter marks in Math: "))
print("Marks in maths entered",math)
science = float(input("Enter marks in Science: "))
print("Marks in science entered",science)
english = float(input("Enter marks in English: "))
print("Marks in english entered",english)

total = math + science + english
percentage = (total / 300) * 100

print("\n----- Result -----")
print("Student Name :", name)
print("Roll Number  :", roll)
print("Percentage   :", round(percentage, 2), "%")

# Finding highest and lowest marks
if math >= science and math >= english:
    highest = "Math"
    high_marks = math
elif science >= math and science >= english:
    highest = "Science"
    high_marks = science
else:
    highest = "English"
    high_marks = english

if math <= science and math <= english:
    lowest = "Math"
    low_marks = math
elif science <= math and science <= english:
    lowest = "Science"
    low_marks = science
else:
    lowest = "English"
    low_marks = english

print("Subject with Highest marks :", highest, "(", high_marks, ")")
print("Subject with Lowest marks  :", lowest, "(", low_marks, ")")
```
