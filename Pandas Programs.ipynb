```python
import pandas as pd
#define a dictonary with sample data which includes some missing values
data={'A':[1,2,3,None,5],'B':[None,2,3,4,5],'C':[1,2,None,None,5]}
df=pd.DataFrame(data)
print("Original Data:\n",df)
print()
#use dropna() to remove rows with any missing values
df_cleaned=df.dropna()
print("Cleaned Data:\n",df_cleaned)
```

```python
import pandas as pd
#define a dictonary with sample data which includes some missing values
data={'A':[1,2,3,None,5],'B':[None,2,3,4,5],'C':[1,2,None,None,5]}
df=pd.DataFrame(data)
print("Original Data:\n",df)
print()
#Filling NaN values with 0
df.fillna(0,inplace=True)
print("Data after filling NaN with 0:\n",df)
```

```python
import pandas as pd
#define a dictonary with sample data which includes some missing values
data={'A':[1,2,3,None,5],'B':[None,2,3,4,5],'C':[1,2,None,None,5]}
df=pd.DataFrame(data)
print("Original Data:\n",df)
print()
#Filling NaN values with mean of each column
df.fillna(df.mean(),inplace=True)
print("Data after filling Nan with mean:\n",df)
```
