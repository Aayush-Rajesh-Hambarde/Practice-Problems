```python
my_dict = {
    101: "Amit",
    102: "Rahul",
    103: "Priya"
}

print("Keys:")
for key in my_dict:
    print(key)

print("\nValues:")
for value in my_dict.values():
    print(value)

print("\nKeys and Values:")
for key, value in my_dict.items():
    print(key, value)
```

```python
my_dict = {"a": 10, "b": 20, "c": 30, "d": 40}

total = 0
for value in my_dict.values():
    total = total + value

print("Sum of values:", total)
```

```python
my_dict = {
    101: "Amit",
    102: "Rahul",
    103: "Priya",
    104: "Sneha",
    105: "Neha"
}

print("Original Student Information:")
print(my_dict)

# Add a new student
my_dict[106] = "Rohan"
print("\nAfter Adding Student:")
print(my_dict)

# Delete a student
del my_dict[103]
print("\nAfter Deleting Student:")
print(my_dict)

# Display student information
print("\nStudent Information:")
for roll_no, name in my_dict.items():
    print("Roll No:", roll_no, "Name:", name)
```

```python
my_dict = {
    "name": "ABC",
    "panel": "B",
    "rollno": 34,
    "marks": [65, 87, 67, 94]
}

# sorted() on a dictionary returns sorted keys
sorted_keys = sorted(my_dict)
print("Sorted keys:", sorted_keys)
```

```python
students = {
    "Amit": 85,
    "Rahul": 92,
    "Priya": 78,
    "Sneha": 95,
    "Neha": 88
}

highest_student = max(students, key=students.get)

print("Student with highest marks:", highest_student)
print("Highest Marks:", students[highest_student])
```

```python
numbers = [10, 20, 10, 30, 20, 10, 40, 30]

frequency = {}

for num in numbers:
    if num in frequency:
        frequency[num] = frequency[num] + 1
    else:
        frequency[num] = 1

print("Frequency of elements:", frequency)
```

```python
my_dict = {
    "Amit": 85,
    "Rahul": 72,
    "Priya": 95,
    "Sneha": 80
}

# Ascending order
sorted_dict = dict(sorted(my_dict.items(), key=lambda item: item[1]))
print("Dictionary sorted by values (Ascending):")
print(sorted_dict)

# Descending order
sorted_dict_desc = dict(sorted(my_dict.items(), key=lambda item: item[1], reverse=True))
print("\nDictionary sorted by values (Descending):")
print(sorted_dict_desc)
```
