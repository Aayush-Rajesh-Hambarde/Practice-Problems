```python
my_list = []
print("Enter 20 values:")
for i in range(20):
    num = int(input(f"Enter value {i+1}: "))
    print("Value",i+1,"entered is:",num)
    my_list.append(num)

print("\nList:", my_list)

# a) Count similar elements and print their index
print("\na) Similar elements and their indices:")
checked = []
for i in range(len(my_list)):
    if my_list[i] not in checked:
        indices = [j for j in range(len(my_list)) if my_list[j] == my_list[i]]
        if len(indices) > 1:
            print(f"Element {my_list[i]} appears at indices: {indices}")
        checked.append(my_list[i])

# b) Count even and odd values
even = 0
odd = 0
for num in my_list:
    if num % 2 == 0:
        even += 1
    else:
        odd += 1
print("\nb) Even values:", even)
print("   Odd values :", odd)

# c) Count positive and negative values
positive = 0
negative = 0
for num in my_list:
    if num > 0:
        positive += 1
    elif num < 0:
        negative += 1
print("\nc) Positive values:", positive)
print("   Negative values:", negative)
```

```python
my_list = []
print("Enter 10 values:")
for i in range(10):
    num = int(input(f"Enter value {i+1}: "))
    print("Value",i+1,"entered is:",num)
    my_list.append(num)

print("\nOriginal list:", my_list)

# a) Sort in ascending order using sorted()
asc_list = sorted(my_list)
print("a) Ascending order (using sorted()):", asc_list)

# b) Sort in descending order using sort()
my_list.sort(reverse=True)
print("b) Descending order (using sort()):", my_list)

# c) Display length of list
print("c) Length of list:", len(my_list))
```

```python
list1 = []
list2 = []

n1 = int(input("How many elements in list1? "))
print("Enter elements of list1:")
for i in range(n1):
    list1.append(int(input()))

n2 = int(input("How many elements in list2? "))
print("Enter elements of list2:")
for i in range(n2):
    list2.append(int(input()))

merged = list1 + list2
print("First list is:",list1)
print("Second list is:",list2)
print("Merged list:", merged)
```

```python
phrase = input("Enter a phrase: ")
print("The phrase is:",phrase)
words = phrase.split()
acronym = ""

for word in words:
    acronym += word[0].upper()

print("Acronym:", acronym)
```

```python
months = ["Jan", "Feb", "Mar", "Apr", "May", "Jun", 
          "Jul", "Aug", "Sep", "Oct", "Nov", "Dec"]

num = int(input("Enter month number (1-12): "))
print("Month number is:",num)

if 1 <= num <= 12:
    print("Abbreviation:", months[num - 1])  # Subtract 1 because list index starts from 0
else:
    print("Invalid month number")
```

```python
keys = ["name", "age", "city"]
values = ["Rahul", 21, "Delhi"]

my_dict = {}
for i in range(len(keys)):
    my_dict[keys[i]] = values[i]

print("Dictionary:", my_dict)
```

```python
my_list = [10, 20, 30, 20, 40, 10, 50]
print("Original list:", my_list)

unique_list = []
for item in my_list:
    if item not in unique_list:
        unique_list.append(item)

print("List after removing duplicates:", unique_list)
```

```python
list1 = [1, 2, 3, 4, 5, 6]
list2 = [4, 5, 6, 7, 8, 9]

common = []
for item in list1:
    if item in list2 and item not in common:
        common.append(item)

print("Common elements:", common)
```
